```python
import matplotlib.pyplot as plt
from wordcloud import WordCloud
from nltk.corpus import inaugural, stopwords
import nltk

nltk.download('stopwords')

# Load a president's inaugural speech
speech = inaugural.raw('2009-Obama.txt')

# Get English stop words
stop_words = set(stopwords.words('english'))

# Remove stop words
words = speech.split()
filtered_words = [word for word in words if word.lower() not in stop_words]

# Join words back into a single string
filtered_text = " ".join(filtered_words)

# Generate the word cloud
wordcloud = WordCloud(
    width=800,
    height=400,
    background_color="white"
).generate(filtered_text)

# Display the word cloud
plt.figure(figsize=(10, 5))
plt.imshow(wordcloud, interpolation="bilinear")
plt.axis("off")
plt.show()
```

```python
from nltk.corpus import inaugural, stopwords
from nltk.probability import FreqDist

# Load the speech
speech = inaugural.raw('2009-Obama.txt')

# Remove stop words and punctuation
stop_words = set(stopwords.words('english'))
words = [
    word.lower()
    for word in speech.split()
    if word.isalpha() and word.lower() not in stop_words
]

# Find word frequencies
freq_dist = FreqDist(words)

# Print the top 5 most frequent words
print(freq_dist.most_common(5))
```

```python
# We are finding the frequency of the word in a text but here we are considering the stopping words also.
text1 = 'Harry goes to a magical school where Harry learns complex spells and Harry makes lifelong friends.Every year, Harry must fight dark forces because Harry is the chosen wizard who must defeat Lord Voldemort.'
text1 = text1.lower()
fd = nltk.FreqDist(text1.split())
```

```python
from nltk.corpus import inaugural
from nltk.corpus import stopwords
from nltk import FreqDist

# Select George Washington's 1789 inaugural speech
file_id = '1789-Washington.txt'

# Preprocess
stop_words = set(stopwords.words('english'))
filtered_words = [
    word.lower()
    for word in inaugural.words(file_id)
    if word.isalpha() and word.lower() not in stop_words
]

# Frequency distribution
fdist = FreqDist(filtered_words)

# Print top 5 words
print("Top 5 high-frequency words:")
for word, count in fdist.most_common(5):
    print(f"{word}: {count}")
```

```python
from nltk.probability import ConditionalFreqDist
cfd = ConditionalFreqDist((len(word),word) for word in text1.split())
cfd
```

```python
from nltk.corpus import inaugural, stopwords
from nltk.probability import FreqDist

# English stop words
stop_words = set(stopwords.words('english'))

# Process each inaugural speech
# Loop through all inaugural speeches
for speech in inaugural.fileids():

    # Get all alphabetic words
    words = [word for word in inaugural.words(speech) if word.isalpha()]

    # Find the longest word
    longest_word = max(words, key=len)

    print(f"{speech} -> {longest_word} ({len(longest_word)} characters)")
```

```python
# Set the required word length to find in each presidents speech

length = 14

for speech in inaugural.fileids():
    words = [
        word.lower()
        for word in inaugural.words(speech)
        if word.isalpha() and len(word) == length
    ]

    # Remove duplicates
    words = sorted(set(words))

    print(f"\n{speech}:")
    print(words)
```
